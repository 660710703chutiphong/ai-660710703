```python
coffee_shop_data = {
    'วันที่': ['1/1/24', '01-01-2024', '2024-01-01', '1 มค 24'],
    'เมนู': ['Latte', 'ลาเต้', 'Latte Hot', 'ลาเต้ร้อน'],
    'ขนาด': ['M', 'กลาง', 'Medium', 'M'],
    'ราคา': [60, 60, None, 60],
    'เวลา': ['08:30', '12.00', '14:30', '9:00']
}
data_frame_coffee_shop = pd.DataFrame(coffee_shop_data)
```

ทำให้ข้อมูลเป็นมาตรฐานเดียวกัน // หาค่าเฉลี่ยของราคา

```python
def Standard_date (day):
    if day in ['1/1/24', '01-01-2024', '2024-01-01', '1 มค 24']:
        return '01/มค/2024'
def Standard_menu(menu):
    if menu in ['Latte', 'ลาเต้', 'Latte Hot', 'ลาเต้ร้อน']:
        return 'Latte Hot 1 Cup'
def Standard_Cup(cup):
    if cup in ['M', 'กลาง', 'Medium', 'M','m']:
        return 'Medium Cup'
def Standard_time(time):
    if pd.isna(time):
        return '08:00'
    time = str(time).replace('.',':')
    if len(time.split(':')[0]) == 1:
        time = '0' + time
    return time;

mean_price = data_frame_coffee_shop['ราคา'].mean()
```

```python
print("="*50)
print("ก่อนแปลงเปลี่ยนค่าต่างๆ")
print("="*50)
print(data_frame_coffee_shop)
print("="*50)
print()
```

```python
data_frame_coffee_shop['ขนาด'] = data_frame_coffee_shop['ขนาด'].apply(Standard_Cup)
data_frame_coffee_shop['วันที่'] = data_frame_coffee_shop['วันที่'].apply(Standard_date)
data_frame_coffee_shop['เวลา'] = data_frame_coffee_shop['เวลา'].apply(Standard_time)
data_frame_coffee_shop['เมนู'] = data_frame_coffee_shop['เมนู'].apply(Standard_menu)

data_frame_coffee_shop['ราคา'].fillna(mean_price,inplace=True)
```

```python
print("="*50)
print("หลังแปลงเปลี่ยนค่าต่างๆ")
print("="*50)
print(data_frame_coffee_shop)
print("="*50)
print()
```

เติมอายุ,ส่วนสูง,น้ำหนัก ที่หายไปด้วยค่าเฉลี่ย

```python
mean_age = df_csv['อายุ'].mean()
mean_height = df_csv['ส่วนสูง_cm'].mean()
mean_weight = df_csv['น้ำหนัก_kg'].mean()
mean_hour = df_csv['ชั่วโมงเรียน_ต่อสัปดาห์'].mean()
df_csv['อายุ'].fillna(mean_age, inplace=True)
df_csv['ส่วนสูง_cm'].fillna(mean_height, inplace=True)
df_csv['น้ำหนัก_kg'].fillna(mean_weight, inplace=True)
df_csv['ชั่วโมงเรียน_ต่อสัปดาห์'].fillna(mean_hour, inplace=True)
```

ทำให้ข้อมูลเป็นมาตรฐานเดียวกัน ร่วมใส่ข้อมูลให้เป็นมาตรฐานเดียวกัน

```python
def standardize_faculty(name):
    if name in ['วิทยาศาสตร์', 'วิทย์', 'Science','Sci', 'คณะวิทยาศาสตร์']:
        return 'วิทยาศาสตร์'
    elif name in ['ศิลปศาสตร์', 'ศิลป์', 'Arts','Liberal Arts']:
        return 'ศิลปศาสตร์'
    elif name in ['วิศวกรรม','วิศวฯ', 'Engineering','วิศวะ', 'วิศวกรรมศาสตร์']:
        return 'วิศวกรรมศาสตร์'
    else:
        return name
df_csv['คณะ'] = df_csv['คณะ'].apply(standardize_faculty)

def standardize_sex(name):
    if name in ['ชาย','ช','M','Male']:
        return 'ชาย'
    elif name in ['หญิง', 'ญ', 'F','Female']:
        return 'หญิง'
    else:
        return name
df_csv['เพศ'] = df_csv['เพศ'].apply(standardize_sex)
```

เติม คณะ,เพศ ที่หายไปโดยอิงจากฐานนิยม

```python
df_csv['คณะ'].fillna(df_csv['คณะ'].mode()[0], inplace=True)
df_csv['เพศ'].fillna(df_csv['เพศ'].mode()[0], inplace=True)
```

```python
Q1 = df_csv['GPA'].quantile(0.25)
Q3 = df_csv['GPA'].quantile(0.75)
IQR = Q3 - Q1
```

จัดการ GPA ที่ผิดปกติ เช่น ติดลบ และเติมเกรดของคนที่ไม่มีเกรด

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# กรองค่าผิดปกติ
df_no_outliers = df_csv[(df_csv['GPA'] >= lower_bound) & (df_csv['GPA'] <= upper_bound)]

print(f"\nขอบเขตล่าง - {lower_bound}")
print(f"ขอบเขตบน - {upper_bound}")
print(f"\nข้อมูลหลังกรองค่าผิดปกติ")
print(df_no_outliers)
print(f"\nค่าเฉลี่ยอายุ (หลังกรองค่าผิดปกติ) - {df_no_outliers['GPA'].mean():.2f}")
```

ลบแถวที่ชื่อเป็น NaN(ไม่ทราบ)

```python
df_master_clean = df_no_outliers.dropna()
df_master_clean
```

รวม Features BMI,สุขภาพ รวมเข้ากับ df_master_clean

```python
df_master_clean['BMI'] = df_master_clean['น้ำหนัก_kg'] / (df_master_clean['ส่วนสูง_cm']/100)**2
df_master_clean['สุขภาพ'] = pd.cut(df_master_clean['BMI'],
                                   bins=[0, 18.5, 25, 30, 100],
                                   labels=['ผอม', 'ปกติ', 'อ้วน', 'อ้วนมาก'])

print(df_master_clean[['ชื่อ', 'BMI', 'สุขภาพ']])
```
