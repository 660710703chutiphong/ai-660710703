```python
# Generate angles from 0 to 2*pi (หน่วยเรเดียน) มีค่าประมาณ 0 ถึง 6.2831
theta = np.linspace(0, 2*np.pi, 1000)

# Calculate x and y coordinates for the unit circle
x = np.cos(theta)
y = np.sin(theta)

fig = go.Figure(data=go.Scattergl(
    x=x,
    y=y,
    mode='lines',
    hoverinfo='text',
    text=[f'x: {xi:.3f}<br>y: {yi:.3f}' for xi, yi in zip(x, y)]
))

# Set layout for the unit circle
fig.update_layout(
    title="Unit Circle with Cosine (x) and Sine (y) Values",
    xaxis_title="X (Cosine)",
    yaxis_title="Y (Sine)",
    xaxis=dict(range=[-1.2, 1.2], constrain="domain"),
    yaxis=dict(scaleanchor="x", scaleratio=1, range=[-1.2, 1.2]),
    hovermode='closest',
    width=600,
    height=600,
)

fig.add_shape(type="circle",
    xref="x", yref="y",
    x0=-1, y0=-1, x1=1, y1=1,
    line_color="LightSeaGreen",
    fillcolor="PaleTurquoise",
    opacity=0.2,
    layer="below"
)


fig.show()
```
